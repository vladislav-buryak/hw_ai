```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split


X = np.array([0, 5, 10, 15, 20, 25], dtype=float).reshape(-1, 1)
y = np.array([45, 41, 36, 31, 27, 24], dtype=float)


def main() -> None:
    print("=" * 62)
    print("ПРОГНОЗ СПОЖИВАННЯ ЕЛЕКТРОЕНЕРГІЇ — ЛІНІЙНА РЕГРЕСІЯ")
    print("=" * 62)

    print("\nВхідні дані:")
    print("Температура (°C) | Споживання (кВт·год)")
    print("-" * 42)
    for temperature, consumption in zip(X.flatten(), y):
        print(f"{temperature:>15.1f} | {consumption:>22.1f}")

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.33, random_state=42
    )

    model = LinearRegression()
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    print("\n" + "=" * 62)
    print("РЕЗУЛЬТАТИ ПЕРЕВІРКИ МОДЕЛІ")
    print("=" * 62)
    print(f"Коефіцієнт нахилу (slope): {model.coef_[0]:.3f}")
    print(f"Вільний член (intercept):  {model.intercept_:.3f}")
    print("\nПорівняння на тестових даних:")
    print("Температура | Фактично | Прогноз | Похибка")
    print("-" * 52)
    for temperature, actual, predicted in zip(
        X_test.flatten(), y_test, y_pred
    ):
        error = abs(actual - predicted)
        print(
            f"{temperature:>10.1f} °C | {actual:>8.1f} | "
            f"{predicted:>7.2f} | {error:>7.2f} кВт·год"
        )

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print("\nМетрики якості:")
    print(f"MAE  — середня абсолютна помилка: {mae:.2f} кВт·год")
    print(f"MSE  — середня квадратична помилка: {mse:.2f}")
    print(f"RMSE — корінь із MSE: {rmse:.2f} кВт·год")
    print(f"R²   — коефіцієнт детермінації: {r2:.3f}")
    print(
        "\nУвага: тестова вибірка містить лише два приклади, тому метрики "
        "є демонстраційними й не дають надійної оцінки для реального будинку."
    )

    print("\n" + "=" * 62)
    while True:
        raw_temperature = input(
            "Введіть середньодобову температуру для прогнозу (°C): "
        ).strip().replace(",", ".")
        try:
            new_temperature = float(raw_temperature)
            break
        except ValueError:
            print("Будь ласка, введіть число, наприклад 12 або 12.5.")

    new_prediction = model.predict(np.array([[new_temperature]]))[0]
    print(
        f"\nЗа температури {new_temperature:.1f} °C модель прогнозує "
        f"{new_prediction:.2f} кВт·год електроенергії на добу."
    )

    if new_temperature < X.min() or new_temperature > X.max():
        print(
            "Примітка: температура поза діапазоном навчальних даних "
            f"({X.min():.0f}…{X.max():.0f} °C). Це екстраполяція, "
            "тому прогноз може бути менш точним."
        )

    temperature_line = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)
    consumption_line = model.predict(temperature_line)

    plt.figure(figsize=(9, 6))
    plt.scatter(
        X_train.flatten(), y_train,
        label="Навчальні дані", marker="o", s=75
    )
    plt.scatter(
        X_test.flatten(), y_test,
        label="Тестові дані", marker="s", s=75
    )
    plt.plot(
        temperature_line.flatten(), consumption_line,
        label="Лінійна регресія", linewidth=2
    )
    plt.scatter(
        [new_temperature], [new_prediction],
        label="Новий прогноз", marker="*", s=180
    )
    plt.title("Прогноз добового споживання електроенергії")
    plt.xlabel("Середньодобова температура, °C")
    plt.ylabel("Споживання електроенергії, кВт·год")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig("energy_regression_plot.png", dpi=150)
    print("\nГрафік збережено у файл: energy_regression_plot.png")
    plt.show()


if __name__ == "__main__":
    main()
```
